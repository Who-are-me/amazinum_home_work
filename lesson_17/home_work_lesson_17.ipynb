```python
# i decided to leave the duplicates!
trans = [grouped.iloc[row].iloc[1] for row in range(grouped.shape[0])]
trans[:5]
```

```python
# i let def settings from example
rules = list(apriori(trans, min_support=0.003, min_confidence=0.2, min_lift=4, min_length=2))
rules
```

```python
# record => items, support, ordered_stat, item_add, confidence, lift
(rules[0])
```

```python
len(rules)
```

```python
# print for details
def print_rules(rules):
    for rule in rules:
        print ('rule.items=', list(rule.items))
        print ('rule.support=',rule.support)

        for os in rule.ordered_statistics:
            print ('\titems_base=', list(os.items_base))
            print ('\tlifted_item =', list(os.items_add))
            print ('\tlift=', os.lift)
            print ('\tconfidence (i.e. cond prob {} if {})='.format(list(os.items_add), list(os.items_base)), os.confidence)
            print ('\n')
```

```python
print_rules(rules)
```

```python
def print_rules_multi_os(rules):
    for rule in rules:
        if len (rule.ordered_statistics)>1:  
            print ('rule.items=', list(rule.items))
            for os in rule.ordered_statistics:
                print ('\n\titems_base=', list(os.items_base))
                print ('\tlifted_item =', list(os.items_add))
                print ('\tlift=', os.lift)
                print ('\tconfidence (i.e. cond prob {} if {})='.format(list(os.items_add), list(os.items_base)), os.confidence)

            break
```

```python
print_rules_multi_os(rules)
```

```python
def get_confidence(rule):
    return max(ordered_stat.confidence for ordered_stat in rule.ordered_statistics)
```

```python
# i select best by confidence parametet
sorted_rules = sorted(rules, key=get_confidence, reverse=True)
```

```python
top_10_rules = sorted_rules[:10]
```

```python
# print top 10 rules by confidence
for rule in top_10_rules:
    items = list(rule.items)
    print(f"Rule: {items}, Support: {rule.support:.2f}")
    for ordered_stat in rule.ordered_statistics:
        base = list(ordered_stat.items_base)
        add = list(ordered_stat.items_add)
        print(f"  {base} => {add}")
        print(f"  Confidence: {ordered_stat.confidence:.2f}, Lift: {ordered_stat.lift:.2f}")
    print()
```
